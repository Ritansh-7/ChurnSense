## Tenure Dist. Analysis

Our tenure histogram has a high concentration of customers at very low tenure and another rise near the maximum tenure.
- Many customers are relatively new specially at 6 starting months .
- There is also a group of long-tenure customers.
- The distribution is not a simple normal or bell-shaped distribution.
This is useful because customer behavior may differ substantially between new and established customers.

## Monthly Charges Dist. Analysis

Our monthly charges histogram has multiple concentrations rather than one smooth peak.

This suggests different pricing or service-bundle groups in the customer base. The lower-charge group may include customers with fewer services, while higher-charge customers may have different service combinations.

We would need to compare service types and bundles to verify this interpretation.

## TotalCharges  Dist. Analysis

Our TotalCharges distribution is strongly right-skewed:
- Many customers have relatively low total charges.
- A smaller group has accumulated much larger charges.
This is expected to be related to tenure because customers who stay longer have more time to accumulate charges.

That relationship is important when we perform feature engineering.

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Create a contingency table
contract_table = pd.crosstab(
    df["Contract"],
    df["Churn"]
)

display(contract_table)

# Perform chi-square test
chi2, p_value, degrees_of_freedom, expected = (
    chi2_contingency(contract_table)
)

print("Chi-square statistic:", chi2)
print("Degrees of freedom:", degrees_of_freedom)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject H0: Contract and churn are associated.")
else:
    print("Fail to reject H0: Insufficient evidence of an association.")
```

```python

import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
from statsmodels.stats.multitest import multipletests

# Work on a copy of the data
stats_df = df.copy()

features_to_test = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod"
]

results = []

for col in features_to_test:
    # Remove rows missing the feature or target
    temp = stats_df[[col, "Churn"]].dropna()

    # Observed frequency table
    table = pd.crosstab(temp[col], temp["Churn"])

    # Chi-square test of independence
    chi2, p_value, dof, expected = chi2_contingency(table)

    # Cramer's V: strength of association
    n = table.to_numpy().sum()
    denominator = min(table.shape[0] - 1, table.shape[1] - 1)

    cramers_v = (
        np.sqrt(chi2 / (n * denominator))
        if denominator > 0 and n > 0
        else np.nan
    )

    results.append({
        "Feature": col,
        "Chi-square": chi2,
        "Degrees of freedom": dof,
        "P-value": p_value,
        "Cramer's V": cramers_v,
        "Min expected frequency": expected.min()
    })

results_df = pd.DataFrame(results)

# Correct for testing multiple hypotheses
results_df["Adjusted p-value"] = multipletests(
    results_df["P-value"],
    method="fdr_bh"
)[1]

results_df["Significant (FDR 5%)"] = (
    results_df["Adjusted p-value"] < 0.05
)

display(
    results_df.sort_values("Cramer's V", ascending=False)
    .round(4)
    .reset_index(drop=True)
)
```

```python

import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
from statsmodels.stats.multitest import multipletests

results = []

for col in features_to_test:
    temp = df[[col, "Churn"]].dropna()
    table = pd.crosstab(temp[col], temp["Churn"])

    chi2, p_value, dof, expected = chi2_contingency(table)

    n = table.to_numpy().sum()
    denominator = min(table.shape[0] - 1, table.shape[1] - 1)

    cramers_v = np.sqrt(chi2 / (n * denominator))

    results.append({
        "Feature": col,
        "Chi-square": chi2,
        "Degrees of freedom": dof,
        "P-value": p_value,
        "Cramer's V": cramers_v,
        "Min expected frequency": expected.min()
    })

results_df = pd.DataFrame(results)

# Apply Benjamini-Hochberg correction to raw p-values
results_df["Adjusted p-value"] = multipletests(
    results_df["P-value"].to_numpy(),
    method="fdr_bh"
)[1]

results_df["Significant (FDR 5%)"] = (
    results_df["Adjusted p-value"] < 0.05
)

# Validate the p-values
assert results_df["Adjusted p-value"].between(0, 1).all()

display(
    results_df.sort_values("Cramer's V", ascending=False)
    .round(6)
    .reset_index(drop=True)
)
```

```python

from sklearn.model_selection import train_test_split

# Separate features and target
X = model_df.drop(columns=["Churn", "customerID"])
y = model_df["Churn"]

# Stratification preserves the churn proportion
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Test features:", X_test.shape)

print(
    "\nTraining churn rate:",
    round(y_train.mean() * 100, 2), "%"
)

print(
    "Test churn rate:",
    round(y_test.mean() * 100, 2), "%"
)

```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

# Identify columns using training data only
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical features:", numeric_features)
print("\nNumber of numerical features:", len(numeric_features))

print("\nCategorical features:", categorical_features)
print("\nNumber of categorical features:", len(categorical_features))
```

```python
# Numerical features: fill missing values, then scale
numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical features: fill missing values, then one-hot encode
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Apply the correct pipeline to each column type
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipeline, numeric_features),
        ("cat", categorical_pipeline, categorical_features)
    ],
    remainder="drop"
)

print("Preprocessing pipeline created successfully.")
```

```python

import pandas as pd

from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

models = {
    "Dummy Baseline": DummyClassifier(strategy="most_frequent"),

    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42,
        class_weight="balanced"
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1
    )
}

results = []
trained_models = {}

for name, model in models.items():
    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_train)
    y_proba = (
        pipeline.predict_proba(X_train)[:, 1]
        if hasattr(pipeline, "predict_proba")
        else None
    )

    results.append({
        "Model": name,
        "Train Accuracy": accuracy_score(y_train, y_pred),
        "Train Precision": precision_score(
            y_train, y_pred, zero_division=0
        ),
        "Train Recall": recall_score(
            y_train, y_pred, zero_division=0
        ),
        "Train F1": f1_score(
            y_train, y_pred, zero_division=0
        ),
        "Train ROC-AUC": (
            roc_auc_score(y_train, y_proba)
            if y_proba is not None else None
        )
    })

    trained_models[name] = pipeline

baseline_results = pd.DataFrame(results)
display(
    baseline_results
    .sort_values("Train F1", ascending=False)
    .round(3)
)
```

```python

from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc",
    "average_precision": "average_precision"
}

cv_results = []

for name, model in models.items():
    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    cv_results.append({
        "Model": name,
        "CV Accuracy": scores["test_accuracy"].mean(),
        "CV Precision": scores["test_precision"].mean(),
        "CV Recall": scores["test_recall"].mean(),
        "CV F1": scores["test_f1"].mean(),
        "CV ROC-AUC": scores["test_roc_auc"].mean(),
        "CV PR-AUC": scores["test_average_precision"].mean(),
        "F1 Std": scores["test_f1"].std()
    })

cv_results_df = pd.DataFrame(cv_results)

display(
    cv_results_df
    .sort_values("CV PR-AUC", ascending=False)
    .round(3)
)
```

```python

from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint

rf_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        random_state=42,
        class_weight="balanced",
        n_jobs=1
    ))
])

param_distributions = {
    "model__n_estimators": randint(150, 501),
    "model__max_depth": [None, 5, 10, 15, 20, 30],
    "model__min_samples_split": randint(2, 21),
    "model__min_samples_leaf": randint(1, 11),
    "model__max_features": ["sqrt", "log2", 0.5]
}

search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_distributions,
    n_iter=25,
    scoring="average_precision",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1,
    refit=True
)

search.fit(X_train, y_train)

print("Best CV PR-AUC:", round(search.best_score_, 4))
print("\nBest parameters:")
print(search.best_params_)
```

```python
from sklearn.model_selection import cross_validate
from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

# Logistic Regression pipeline
logistic_pipeline = Pipeline(steps=[
    ("preprocessor", clone(preprocessor)),
    ("model", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

# Best tuned Random Forest: search.best_estimator_ already
# contains the fitted preprocessing/model pipeline.
# Clone it to evaluate fairly from scratch in each CV fold.
tuned_rf_pipeline = clone(search.best_estimator_)

candidate_models = {
    "Logistic Regression": logistic_pipeline,
    "Random Forest (default)": Pipeline(steps=[
        ("preprocessor", clone(preprocessor)),
        ("model", RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            class_weight="balanced",
            n_jobs=1
        ))
    ]),
    "Random Forest (tuned)": tuned_rf_pipeline
}

comparison = []

for name, pipeline in candidate_models.items():
    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring={
            "precision": "precision",
            "recall": "recall",
            "f1": "f1",
            "roc_auc": "roc_auc",
            "pr_auc": "average_precision"
        },
        n_jobs=-1
    )

    comparison.append({
        "Model": name,
        "CV Precision": scores["test_precision"].mean(),
        "CV Recall": scores["test_recall"].mean(),
        "CV F1": scores["test_f1"].mean(),
        "CV ROC-AUC": scores["test_roc_auc"].mean(),
        "CV PR-AUC": scores["test_pr_auc"].mean(),
        "F1 Std": scores["test_f1"].std()
    })

comparison_df = pd.DataFrame(comparison)

display(
    comparison_df
    .sort_values("CV PR-AUC", ascending=False)
    .round(3)
    .reset_index(drop=True)
)
```

```python
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import (
    precision_recall_curve,
    precision_score,
    recall_score,
    f1_score
)

# Use the tuned Random Forest candidate
rf_candidate = clone(search.best_estimator_)

oof_probabilities = cross_val_predict(
    rf_candidate,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)[:, 1]

print("OOF predictions:", len(oof_probabilities))
print("Training rows:", len(y_train))
```

```python
precision, recall, thresholds = precision_recall_curve(
    y_train,
    oof_probabilities
)

# The last precision/recall point has no corresponding threshold
f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-12)
)

best_index = np.argmax(f1_scores)
best_threshold = thresholds[best_index]

print("Selected threshold:", round(best_threshold, 3))
print("OOF precision:", round(precision[best_index], 3))
print("OOF recall:", round(recall[best_index], 3))
print("OOF F1:", round(f1_scores[best_index], 3))
```

# ChurnSense — Model Card

## Project objective

Estimate the likelihood of customer churn using customer account and service information. The intended use is analytical decision support for identifying customers who may warrant further review.

## Dataset

- Domain: Telecom customer churn
- Dataset size: 7,043 rows and 21 original columns
- Target: Customer churn
- Split: 5,634 training rows and 1,409 test rows
- Test-set churn rate: approximately 26.54%

## Model development

- Preprocessing: Median imputation and scaling for numerical features; most-frequent imputation and one-hot encoding for categorical features.
- Models evaluated: Dummy classifier, Logistic Regression, Decision Tree, and Random Forest.
- Validation: Stratified cross-validation.
- Hyperparameter tuning: RandomizedSearchCV.
- Decision threshold: 0.618, selected using out-of-fold predictions.

## Reported test performance

- ROC-AUC: 0.841
- PR-AUC (Average Precision): 0.647
- Accuracy: 0.78
- Churn precision: 0.57
- Churn recall: 0.64
- Churn F1-score: 0.61

These results describe performance on the held-out test set, not guaranteed future performance.

## Intended use

- Exploratory churn-risk analysis.
- Prioritization for human review.
- Evaluating possible retention strategies.

## Limitations

- The dataset may not represent the current customer population.
- Predicted probabilities may require calibration.
- False positives and false negatives have different business costs.
- The model does not establish why a customer churns or whether a retention offer will prevent churn.
- The model's performance may change over time.

## Before operational deployment

- Validate data quality and feature definitions.
- Check performance across meaningful customer segments.
- Establish probability calibration and monitoring procedures.
- Validate the model with more recent, representative data.
- Evaluate retention interventions experimentally.
- Document model versions, dependencies, threshold rationale, and approval for use.


```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    brier_score_loss,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report,
)

# --------------------------------------------------
# 1. Validate that the required variables exist
# --------------------------------------------------

required_variables = [
    "final_model",
    "X_test",
    "y_test",
]

missing_variables = [
    name for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        f"Missing notebook variables: {missing_variables}. "
        "Run your model-training and test-split cells first."
    )

THRESHOLD = 0.618

if not 0 <= THRESHOLD <= 1:
    raise ValueError("Threshold must be between 0 and 1.")

if len(X_test) != len(y_test):
    raise ValueError("X_test and y_test have different row counts.")

if len(X_test) == 0:
    raise ValueError("The test set is empty.")


# --------------------------------------------------
# 2. Identify the churn class correctly
# --------------------------------------------------

model_classes = list(final_model.classes_)

positive_candidates = [
    label for label in model_classes
    if (
        label == 1
        or str(label).strip().lower()
        in {"yes", "true", "churn", "churned"}
    )
]

if len(positive_candidates) != 1:
    raise ValueError(
        f"Cannot identify the churn class unambiguously. "
        f"Model classes: {model_classes}. "
        "Check how the target variable was encoded."
    )

positive_label = positive_candidates[0]
positive_index = model_classes.index(positive_label)


# --------------------------------------------------
# 3. Generate test-set probabilities
# --------------------------------------------------

y_probability = final_model.predict_proba(X_test)[:, positive_index]

# Convert test labels to a consistent binary representation.
y_actual = (
    pd.Series(y_test).reset_index(drop=True) == positive_label
).astype(int).to_numpy()

if not np.isfinite(y_probability).all():
    raise ValueError("Model probabilities contain NaN or infinity.")

if not np.isin(y_actual, [0, 1]).all():
    raise ValueError("Unexpected target labels.")

if len(np.unique(y_actual)) != 2:
    raise ValueError("Both churn classes must be present in the test set.")

if not np.isfinite(y_probability).all() or (
    (y_probability < 0) | (y_probability > 1)
).any():
    raise ValueError("Invalid probability values detected.")

y_predicted = (
    y_probability >= THRESHOLD
).astype(int)


# --------------------------------------------------
# 4. Calculate final evaluation metrics
# --------------------------------------------------

tn, fp, fn, tp = confusion_matrix(
    y_actual,
    y_predicted,
    labels=[0, 1],
).ravel()

test_metrics = pd.DataFrame({
    "Metric": [
        "Test customers",
        "Actual churn prevalence",
        "ROC-AUC",
        "PR-AUC (Average Precision)",
        "Accuracy",
        "Churn precision",
        "Churn recall",
        "Churn F1-score",
        "Brier score",
        "True negatives",
        "False positives",
        "False negatives",
        "True positives",
        "Customers flagged",
        "Flagged customer rate",
    ],
    "Value": [
        len(y_actual),
        y_actual.mean(),
        roc_auc_score(y_actual, y_probability),
        average_precision_score(y_actual, y_probability),
        accuracy_score(y_actual, y_predicted),
        precision_score(y_actual, y_predicted, zero_division=0),
        recall_score(y_actual, y_predicted, zero_division=0),
        f1_score(y_actual, y_predicted, zero_division=0),
        brier_score_loss(y_actual, y_probability),
        tn,
        fp,
        fn,
        tp,
        int(y_predicted.sum()),
        y_predicted.mean(),
    ],
})

display(test_metrics.round(4))

print(f"\nSelected threshold: {THRESHOLD:.3f}")
print(f"Model classes: {model_classes}")
print(f"Churn class: {positive_label}")

print("\nClassification report:")
print(
    classification_report(
        y_actual,
        y_predicted,
        labels=[0, 1],
        target_names=["No Churn", "Churn"],
        zero_division=0,
    )
)


# --------------------------------------------------
# 5. Confusion matrix
# --------------------------------------------------

ConfusionMatrixDisplay.from_predictions(
    y_actual,
    y_predicted,
    labels=[0, 1],
    display_labels=["No Churn", "Churn"],
    values_format="d",
)

plt.title(f"ChurnSense Test Results — Threshold {THRESHOLD}")
plt.tight_layout()
plt.show()


# --------------------------------------------------
# 6. Compare PR-AUC against a prevalence baseline
# --------------------------------------------------

baseline_pr_auc = y_actual.mean()
model_pr_auc = average_precision_score(
    y_actual,
    y_probability,
)

print(f"Churn prevalence baseline: {baseline_pr_auc:.4f}")
print(f"Model average precision:   {model_pr_auc:.4f}")
print(
    "PR-AUC lift over prevalence baseline: "
    f"{model_pr_auc / baseline_pr_auc:.2f}x"
)
```

```python

# --------------------------------------------------
# Bootstrap confidence intervals
# --------------------------------------------------

rng = np.random.default_rng(42)
N_BOOTSTRAPS = 1000

negative_idx = np.flatnonzero(y_actual == 0)
positive_idx = np.flatnonzero(y_actual == 1)

roc_bootstrap = []
pr_bootstrap = []

for _ in range(N_BOOTSTRAPS):

    # Stratified resampling preserves both class counts.
    sampled_negative = rng.choice(
        negative_idx,
        size=len(negative_idx),
        replace=True,
    )

    sampled_positive = rng.choice(
        positive_idx,
        size=len(positive_idx),
        replace=True,
    )

    sample_idx = np.concatenate([
        sampled_negative,
        sampled_positive,
    ])

    actual_sample = y_actual[sample_idx]
    probability_sample = y_probability[sample_idx]

    roc_bootstrap.append(
        roc_auc_score(actual_sample, probability_sample)
    )

    pr_bootstrap.append(
        average_precision_score(
            actual_sample,
            probability_sample,
        )
    )


def bootstrap_summary(name, point_estimate, samples):
    return {
        "Metric": name,
        "Estimate": point_estimate,
        "95% CI Lower": np.percentile(samples, 2.5),
        "95% CI Upper": np.percentile(samples, 97.5),
    }


confidence_intervals = pd.DataFrame([
    bootstrap_summary(
        "ROC-AUC",
        roc_auc_score(y_actual, y_probability),
        roc_bootstrap,
    ),
    bootstrap_summary(
        "PR-AUC",
        average_precision_score(y_actual, y_probability),
        pr_bootstrap,
    ),
])

display(confidence_intervals.round(4))
```


## Segment-Level Model Validation

### Objective
Evaluate whether churn prevalence, model recall, and false-positive rates vary across customer segments.

### Important
These results are diagnostic. Segment-level differences do not automatically imply bias or justify changing the model.


```python

import numpy as np
import pandas as pd
from sklearn.metrics import recall_score, precision_score, f1_score

segment_data = X_test.reset_index(drop=True).copy()

segment_data["Actual_Churn"] = np.asarray(y_actual)
segment_data["Churn_Probability"] = np.asarray(y_probability)
segment_data["Predicted_Churn"] = (
    segment_data["Churn_Probability"] >= THRESHOLD
).astype(int)

# Analyze available categorical customer features.
categorical_features = segment_data.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Include binary numerical categories such as SeniorCitizen.
for col in X_test.select_dtypes(include=["number"]).columns:
    unique_values = set(X_test[col].dropna().unique())
    if unique_values.issubset({0, 1}) and len(unique_values) == 2:
        if col not in categorical_features:
            categorical_features.append(col)

rows = []

for feature in categorical_features:
    for segment, group in segment_data.groupby(
        feature, dropna=False
    ):
        if len(group) < 30:
            continue

        actual = group["Actual_Churn"].to_numpy()
        predicted = group["Predicted_Churn"].to_numpy()

        tp = int(((actual == 1) & (predicted == 1)).sum())
        fp = int(((actual == 0) & (predicted == 1)).sum())
        fn = int(((actual == 1) & (predicted == 0)).sum())
        tn = int(((actual == 0) & (predicted == 0)).sum())

        rows.append({
            "Feature": feature,
            "Segment": str(segment),
            "Customers": len(group),
            "Actual Churn Rate (%)": 100 * actual.mean(),
            "Flagged Rate (%)": 100 * predicted.mean(),
            "Recall": recall_score(
                actual, predicted, zero_division=0
            ) if actual.sum() > 0 else np.nan,
            "Precision": precision_score(
                actual, predicted, zero_division=0
            ),
            "False Positive Rate": (
                fp / (fp + tn) if (fp + tn) > 0 else np.nan
            ),
            "Missed Churners": fn,
        })

segment_results = pd.DataFrame(rows)

if segment_results.empty:
    print("No eligible segments found. Check the feature dtypes.")
else:
    display(
        segment_results.sort_values(
            ["Feature", "Customers"],
            ascending=[True, False]
        ).round(3)
    )
```

```python

import joblib
import pandas as pd
import streamlit as st
from pathlib import Path

st.set_page_config(
    page_title="ChurnSense",
    page_icon="📊",
    layout="wide"
)

ARTIFACT_PATH = Path(r"D:\Webots\churnsense_artifact.joblib")
artifact = joblib.load(ARTIFACT_PATH)

model = artifact["model"]
threshold = artifact["threshold"]
feature_columns = artifact["feature_columns"]

st.title("📊 ChurnSense")
st.subheader("Telecom Customer Churn Prediction")
st.write(
    "Estimate churn risk from customer details "
    "and identify customers for retention outreach."
)

st.info(
    f"Current classification threshold: {threshold:.3f}. "
    "Predictions are estimates, not guarantees."
)

# Create one input for every model feature.
# Numeric and categorical values must retain their original types.
customer = {}

with st.form("customer_form"):
    st.markdown("### Customer details")

    for feature in feature_columns:
        if feature in ["tenure", "MonthlyCharges", "TotalCharges"]:
            customer[feature] = st.number_input(
                feature,
                min_value=0.0,
                value=12.0 if feature == "tenure" else 70.0
            )
        else:
            customer[feature] = st.text_input(
                feature,
                value=""
            )

    submitted = st.form_submit_button("Predict churn risk")

if submitted:
    customer_df = pd.DataFrame([customer], columns=feature_columns)

    try:
        probability = float(model.predict_proba(customer_df)[0, 1])
        predicted_churn = probability >= threshold

        col1, col2 = st.columns(2)

        col1.metric("Estimated churn probability", f"{probability:.1%}")
        col2.metric(
            "Predicted class",
            "Churn risk" if predicted_churn else "Lower risk"
        )

        if predicted_churn:
            st.warning(
                "This customer is flagged for potential retention outreach."
            )
        else:
            st.success(
                "This customer is below the current classification threshold."
            )

        st.caption(
            "Validate input values against the training dataset's allowed "
            "categories before using this dashboard operationally."
        )

    except Exception as exc:
        st.error(f"Prediction failed: {exc}")
```
